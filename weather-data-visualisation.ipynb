```python
# Import all necessary files and create variables

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import os
from ipywidgets import interact, Dropdown
import seaborn as sns
import numpy as np

final_weather_data_folder = '../data'
```

```python
# Create a list of file names ending in '_weather_data.csv'
csv_files = [file for file in os.listdir(final_weather_data_folder) if file.endswith('_weather_data.csv')]
print(csv_files)


# Read all CSV files into DataFrames and store them in a list
dfs = [pd.read_csv(os.path.join(final_weather_data_folder, csv_file)) for csv_file in csv_files]

weather_df = pd.concat(dfs, ignore_index=True)
```

```python
# Verify df
print(weather_df)
```

```python
# Rename variables for consistency and print
weather_df.rename(columns={'Date' : 'DATE', 'TMAX (°C)' : 'TMAX', 'TMIN (°C)' : 'TMIN'}, inplace=True)
print(weather_df)
```

```python
# Do processing for visualization
weather_df['DATE'] = pd.to_datetime(weather_df['DATE'], errors='coerce')  # Ensure DATE is correctly formatted

# Drop rows with missing or NaN values in important columns
weather_df_cleaned = weather_df.dropna(subset=['TMAX', 'TMIN', 'DATE', 'STATE'])

# Remove rows where TMAX is smaller than TMIN
weather_df_cleaned = weather_df_cleaned[weather_df_cleaned['TMAX'] >= weather_df_cleaned['TMIN']]

# Remove rows where TMAX is hotter than the hottest recorded day in America (56.7°C)
weather_df_cleaned = weather_df_cleaned[weather_df_cleaned['TMAX'] <= 56.7]

# Remove rows with temperatures below an extreme threshold (e.g., colder than -90°C)
weather_df_cleaned = weather_df_cleaned[weather_df_cleaned['TMAX'] >= -90]

# Remove duplicates
weather_df_cleaned = weather_df_cleaned.drop_duplicates()

# Check if the rows were successfully removed
print(f"Removed {len(weather_df) - len(weather_df_cleaned)} rows during cleaning")

# Update weather_df to only include cleaned data
weather_df = weather_df_cleaned

# Display the cleaned DataFrame
print(weather_df)
```

```python
# PLOTS AVERAGE TEMPERATURE FOR EACH STATE

# Extract month from date
weather_df['MONTH'] = weather_df['DATE'].dt.month

# Group by State and Month, then calculate the average TMAX
monthly_TMAX = weather_df.groupby(['STATE', 'MONTH'])['TMAX'].mean().reset_index()

# Function to plot data for selected state
def plot_state_temperature(selected_state):
    plt.figure(figsize=(26, 12))

    if selected_state == 'All States':
            states = monthly_TMAX['STATE'].unique()
            for state in states:
                state_data = monthly_TMAX[monthly_TMAX['STATE'] == state]
                plt.plot(state_data['MONTH'], state_data['TMAX'], marker='o', label=state)
    else:
        state_data = monthly_TMAX[monthly_TMAX['STATE'] == selected_state]
        plt.plot(state_data['MONTH'], state_data['TMAX'], marker='o', label=selected_state)
    
    plt.title(f'Monthly Average TMAX for {selected_state}')
    plt.xlabel('Month')
    plt.ylabel('Average TMAX (°C)')
    plt.xticks(range(1, 13), ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
    plt.ylim(-10, 40)
    plt.xlim(0.5, 12.5)
    plt.grid(True)
    plt.legend(title='State')
    plt.tight_layout()
    plt.show()

# Create a dropdown menu for state selection
state_dropdown = Dropdown(
    options=['All States'] + list(monthly_TMAX['STATE'].unique()),
    description='Select State:',
    value='All States'  # Default value
)

# Create the interactive plot
interact(plot_state_temperature, selected_state=state_dropdown)
```

```python
# Scatter plot to identify uncleaned data

# Extract the day from the DATE column
weather_df['Day'] = weather_df['DATE'].dt.dayofyear

# Function to plot data for selected state
def plot_state_temperature(selected_state):
    plt.figure(figsize=(26, 12))
    
    state_data = weather_df[weather_df['STATE'] == selected_state]
    plt.scatter(state_data['Day'], state_data['TMAX'], label=selected_state, s=100, alpha=0.6)  # Scatter plot for selected state
    
    plt.title(f'Daily TMAX for {selected_state}')
    plt.xlabel('Day of the Year')
    plt.ylabel('TMAX')
    plt.ylim(-50, 70)
    plt.xlim(0, 365)
    plt.grid(True)
    
    plt.tight_layout()
    plt.show()

# Create a dropdown menu for state selection
state_dropdown = Dropdown(
    options=list(weather_df['STATE'].unique()),  # Only list unique states, no 'All States'
    description='Select State:',
    value=list(weather_df['STATE'].unique())[0]
)

# Create the interactive plot
interact(plot_state_temperature, selected_state=state_dropdown)
```

```python
# Convert DATE column to datetime
weather_df['DATE'] = pd.to_datetime(weather_df['DATE'])

# Extract the day of the year and the year
weather_df['Day'] = weather_df['DATE'].dt.dayofyear
weather_df['Year'] = weather_df['DATE'].dt.year

# Function to plot data for selected state and year
def plot_state_temperature(selected_state, selected_year):
    plt.figure(figsize=(26, 12))
    
    # Filter data for the selected state and year
    state_data = weather_df[(weather_df['STATE'] == selected_state) & (weather_df['Year'] == selected_year)]
    
    # Scatter plot for selected state and year
    plt.scatter(state_data['Day'], state_data['TMAX'], label=f'{selected_state} ({selected_year})', s=100, alpha=0.6)
    
    plt.title(f'Daily TMAX for {selected_state} in {selected_year}')
    plt.xlabel('Day of the Year')
    plt.ylabel('TMAX (°C)')
    plt.ylim(-50, 70)
    plt.xlim(0, 365)
    plt.grid(True)
    
    plt.tight_layout()
    plt.show()

# Create dropdown menus for state and year selection
state_dropdown = Dropdown(
    options=list(weather_df['STATE'].unique()),  # List unique states
    description='Select State:',
    value=list(weather_df['STATE'].unique())[0]  # Default value
)

year_dropdown = Dropdown(
    options=sorted(weather_df['Year'].unique()),  # List unique years, sorted
    description='Select Year:',
    value=sorted(weather_df['Year'].unique())[0]  # Default value
)

# Create the interactive plot with both dropdowns
interact(plot_state_temperature, selected_state=state_dropdown, selected_year=year_dropdown)
```

```python
# Easier to Identify outliers with seaborn


# Function to plot a boxplot for selected state
def plot_state_boxplot(selected_state):
    plt.figure(figsize=(26, 12))
    
    state_data = weather_df[weather_df['STATE'] == selected_state]
    sns.boxplot(x=state_data['Day'], y=state_data['TMAX'], color='skyblue')
    
    plt.title(f'TMAX Distribution for {selected_state}')
    plt.xlabel('Day of the Year')
    plt.ylabel('TMAX')
    plt.ylim(-50, 70)
    plt.xlim(0, 365)
    plt.grid(True)
    
    plt.tight_layout()
    plt.show()

interact(plot_state_boxplot, selected_state=state_dropdown)
```

```python
# Frequency of temperatures

# Function to plot a histogram for selected state
def plot_state_histogram(selected_state):
    plt.figure(figsize=(18, 6))  # Width and height of the figure
    
    state_data = weather_df[weather_df['STATE'] == selected_state]
    
    # Define bin edges from -20 to 60 in 5-degree increments
    bin_edges = np.arange(-20, 65, 5)  # Creates an array: [-20, -15, -10, ..., 60]
    
    plt.hist(state_data['TMAX'], bins=bin_edges, color='skyblue', edgecolor='black', alpha=0.7)  # Use defined bins
    
    plt.title(f'TMAX Distribution for {selected_state}')
    plt.xlabel('TMAX (Temperature)')
    plt.ylabel('Frequency')
    plt.grid(True)
    
    # Set x-axis limits for temperature range
    plt.xlim(-20, 60)  # Limit temperature range from -20 to 60
    plt.xticks(range(-20, 61, 5))  # Set x-axis ticks in 5-degree increments
    
    plt.tight_layout()
    plt.show()

# Create the interactive plot with dropdown for histogram
interact(plot_state_histogram, selected_state=state_dropdown)
```
